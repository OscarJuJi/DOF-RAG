### Metrica BLEU

```python
hypothesis = dof_note1
reference = summary1
BLEUscore = nltk.translate.bleu_score.sentence_bleu([reference], hypothesis)
print(f"BLEUscore para la nota 1 : {BLEUscore}")
```

```python
hypothesis = dof_note2
reference = summary2
BLEUscore = nltk.translate.bleu_score.sentence_bleu([reference], hypothesis)
print(f"BLEUscore para la nota 2 : {BLEUscore}")
```

### Metrica Rouge

```python
from rouge_score import rouge_scorer
```

```python
# Initialize the scorer
scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL', 'rougeLsum'])

# Compute the Rouge scores for reference and candidate.
scores = scorer.score(dof_note1,
                      summary1)

# Print the scores
print("Scores para la nota 1:")
scores
```

```python
# Initialize the scorer
scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL', 'rougeLsum'])

# Compute the Rouge scores for reference and candidate.
scores = scorer.score(dof_note2,
                      summary2)

# Print the scores
print("Scores para la nota 2:")
scores
```

### Explicación:
Ambos resultados de las métricas son bastante aceptables para los resúmenes creados por el modelo de IA.

* BLEU

Nos indica que el modelo está al nivel humano para crear un resumen, ya que el resultado es de 0.29, donde un valor más cercano a 1 es mejor. Sin embargo, no es posible alcanzar un valor de 1 en esta métrica, y los resúmenes hechos por una persona normal rondan entre 0.3 y 0.5, por lo que este resultado es bueno.

* ROUGE


Nos proporciona tres métricas: precisión, recall y media armónica.

En cuanto a la precisión, podemos observar que el número de oraciones encontradas pertenece, en general, al texto de la nota de la cual se extrajo la información. Por lo tanto, casi no hubo alucinaciones, y al leer el resumen, este concuerda con la nota.

Podemos observar una media armónica aceptable, porque el modelo resume adecuadamente la información obtenida de la nota sin copiar exactamente las mismas palabras, manteniendo la coherencia del texto.

* Notas Adicionales:

La precisión mide cuántas de las palabras o frases generadas por el modelo aparecen en el texto de referencia.
El recall mide cuántas de las palabras o frases del texto de referencia aparecen en el resumen generado por el modelo.
La media armónica (F1 score) es una combinación de precisión y recall, ofreciendo una medida equilibrada de ambas.
